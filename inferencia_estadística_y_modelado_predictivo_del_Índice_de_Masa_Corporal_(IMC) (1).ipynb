Actividad 1

Se cree que las horas de ejercicio físico semanal (HorasEjercicio) promedio de la población de los participantes es significativamente menor a 8 horas. Realicé el test de hipótesis que permite evaluar la afirmación anterior indicando claramente la hipótesis nula y alternativa, el estadístico observado y el p_valor obtenido, con estos datos responder explícitamente al cuestionamiento realizado.

# Hipótesis:
Hipótesis nula (H0):* La media de HorasEjercicio es mayor o igual a 8. (μ >= 8)
Hipótesis alternativa (H1):* La media de HorasEjercicio es menor que 8. (μ < 8)

```python
t_statis, p_valor = stats.ttest_1samp(data['HorasEjercicio'], popmean=8)
p_valor_unilateral = p_valor / 2
print(f"Estadístico t observado: {t_statis}")
print(f"P-valor unilateral: {p_valor_unilateral}")

# Nivel de significancia
alpha = 0.05
```

Con un estadístico observado t=−0,32 y un p-valor unilateral de 0,37 (mayor a α=0,05), no existe evidencia suficiente para rechazar la hipótesis nula. Por lo tanto, no se puede afirmar que el promedio de horas de ejercicio sea significativamente menor a 8 horas semanales.

Actividad 2

Determinar si existe una correlación significativa entre las variables IMC y Edad, realizar un test de hipótesis de correlación de Pearson para esto. Dentro del análisis incluir claramente las hipótesis nula y alternativa, la correlación obtenida y el p_valor.

Hipótesis para la correlación entre IMC y Edad:
Hipótesis nula (H0): No existe correlación lineal entre IMC y Edad. (ρ = 0)
Hipótesis alternativa (H1):Existe correlación lineal entre IMC y Edad. (ρ ≠ 0)

```python
# Correlación de Pearson y el p-valor
correlacion, p_valor_correlacion = stats.pearsonr(data['IMC'], data['Edad'])
print(f"Correlación de Pearson entre IMC y Edad: {correlacion}")
print(f"P-valor del test de correlación: {p_valor_correlacion}")

# Nivel de significancia
alpha = 0.05
```

En base a los resultados, el p-valor obtenido (p<0,001) es significativamente menor al nivel de significancia del 5% (α=0,05), por lo que se rechaza la hipótesis nula.

Esto indica que existe una correlación lineal positiva estadísticamente significativa entre el IMC y la edad, asociando un mayor índice de masa corporal a medida que aumenta la edad.

En base al modelo obtenido realizar:

d) Determinar si todas las variables independientes son significativas para el modelo, si es necesario eliminar la o las variables no significativas y entregar la ecuación del nuevo modelo, además comparar el R cuadrado de ambos modelos.

α  = 0.05

```python
#a) Realizar el modelo de regresión lineal múltiple para predecir la variable
# IMC en función de las variables Edad, HorasEjercicio y ConsumoCalorias.
# Variables dependiente e independientes
y = data['IMC']
X = data[['Edad', 'HorasEjercicio', 'ConsumoCalorias']]

# Constante para el intercepto
X = sm.add_constant(X)

# Modelo de regresión lineal múltiple
model = sm.OLS(y, X).fit()

# Imprimir el resumen del modelo
print(model.summary())
```

```python
# b) Entregar la ecuación del modelo obtenido e interpretar los coeficientes.
(f"IMC = {model.params['const']:.2f} + {model.params['Edad']:.2f}* Edad\
+ {model.params['HorasEjercicio']:.2f} * HorasEjercicio +\
{model.params['ConsumoCalorias']:.4f} * ConsumoCalorias")
```

Ecuación del modelo

IMC = -0.2130 + 0.2485 * Edad + -0.1494 * HorasEjercicio + 0.0001 * ConsumoCalorias.

-0.21, es el valor de IMC que predice si las horasEjercicio, edad y consumoCalorias fueran cero. no teniendo sentido en la vida real, ya que no puede ser cero.

0,25 (Edad): "Por cada año adicional de edad, el IMC estimado aumenta en promedio 0,25 puntos de IMC (manteniendo las demás variables constantes)."

0,15 (HorasEjercicio): "Por cada hora adicional de ejercicio semanal, el IMC estimado disminuye en promedio 0,15 puntos de IMC (ceteris paribus)."

0,0001 (ConsumoCalorias): "Por cada caloría diaria adicional consumida, el IMC aumenta en promedio 0,0001 puntos, reflejando un impacto marginal."


c) Realizar un análisis general del modelo considerando el R cuadrado y el test de linealidad.

EL 92.9% de los cambios en el IMC se explican por los cambios en la edad, horas de ejercicio y consumo de calorias. Lo que nos muestra la importancia de estos factores.

El p-valor nos indica que es mucho menor al 5% de significancia, lo que nos dice que el modelo nos sirve para aclarar que al menos una de estas variables tienen una relacion real con el IMC

```python
#d) Determinar si todas las variables independientes son significativas para
#el modelo, si es necesario eliminar la o las variables no significativas y
#entregar la ecuación del nuevo modelo, además comparar el R cuadrado de
#ambos modelos

# Variables dependiente e independientes
y = data['IMC']
X = data[['Edad', 'HorasEjercicio', 'ConsumoCalorias']]

# Constante para el intercepto
X = sm.add_constant(X)

# Modelo de regresión lineal múltiple
model = sm.OLS(y, X).fit()

# R-cuadrado del modelo original
r_squared_original = model.rsquared

# Determinar variables significativas (p-valor < 0.05)
significant_vars = model.pvalues[model.pvalues < 0.05].index.tolist()
if 'const' in significant_vars:
    significant_vars.remove('const')

# Si hay variables no significativas, construir un nuevo modelo
if len(significant_vars) < X.shape[1] - 1:
    X_new = data[significant_vars]
    X_new = sm.add_constant(X_new)
    model_new = sm.OLS(y, X_new).fit()
    r_squared_new = model_new.rsquared
else:
    model_new = model
    r_squared_new = r_squared_original

# Imprimir la ecuación del nuevo modelo si fue modificado
if model_new is not model:
    print("Nuevo modelo (variables significativas):")
    equation_new = f"IMC = {model_new.params['const']:.2f}"
    for var in significant_vars:
        equation_new += f" + {model_new.params[var]:.2f} * {var}"
    print(equation_new)

print(model_new.summary())
```

## *Aqui validamos si cada variable es relevante para el modelo*

P-valores individuales de las variables (del primer modelo):

Edad: p-valor es 0.000 (muy significativo)
HorasEjercicio: p-valor es 0.000 (muy significativo)
ConsumoCalorias: p-valor es 0.671 (no significativo)
Conclusión de significancia:

Edad y HorasEjercicio son muy importantes para predecir el IMC (sus p-valores son menores a 0.05).
ConsumoCalorias no es estadísticamente importante para predecir el IMC en este modelo (su p-valor 0.671 es mayor a 0.05).
Nuevo Modelo (después de eliminar ConsumoCalorias):
Como ConsumoCalorias no era significativa, se construyó un nuevo modelo sin ella.

Ecuación del nuevo modelo:
IMC=−0.09+0.25×Edad−0.15×HorasEjercicio

*Comparación de R cuadrado:

R² del modelo original: 0.9293
 R² del nuevo modelo: 0.9292

* El R cuadrado apenas cambió (bajó muy poquito). Esto confirma que quitar la variable ConsumoCalorias no afectó la capacidad del modelo para explicar el IMC de manera significativa, porque no era una variable importante desde el principio. El nuevo modelo es más simple y casi igual de bueno.*

```python
#e) Estudiar la normalidad de los residuos a través del test de Shapiro,
#en la conclusión indicar claramente las hipótesis nula y alternativa,
#y el p_valor obtenido.
residuals = model.resid
shapiro_test_statistic, shapiro_p_value = stats.shapiro(residuals)
print(f"- Estadístico de Shapiro-Wilk: {shapiro_test_statistic:.2f}")
print(f"- P-valor obtenido: {shapiro_p_value:.2f}")
```

Hipótesis Nula (H0): Los residuos siguen una distribución normal.
Hipótesis Alternativa (H1): Los residuos no siguen una distribución normal.


Estadístico de Shapiro-Wilk: 0, 99

ya que el p-valor (0.77) es mayor que el nivel de significancia (0.05), no rechazamos la hipótesis nula.
No existe evidencia estadística suficiente para concluir que los residuos NO se distribuyen de forma normal. por lo que esto nos indica que si siguen una distribucion normal, lo cual valida el modelo de regresion.
